```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")          # a temporary database, gone when you close it

conn.executescript("""
CREATE TABLE sales (
    sale_id   INTEGER PRIMARY KEY,
    item      TEXT,
    quantity  INTEGER
);
""")

toy = pd.DataFrame({"item": ["cup", "bowl", "fork"], "quantity": [6, 3, 10]})
toy.to_sql("sales", conn, if_exists="append", index=False)

pd.read_sql_query("SELECT * FROM sales", conn)
```
